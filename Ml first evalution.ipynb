## import Dataset

```python
import yfinance as yf

stock = input("Enter stock symbol (Ex. TCS, RELIANCE, INFY): ").upper()

stock += ".NS"

print("Downloading data.....")

try:
    df = yf.download(stock, period="10y", progress=False)

    if df.empty:
        print("Please enter a valid stock symbol. Data not found!")
    else:
        print("Download successful!\n")

        print(df.sort_index(ascending=False))
        filename = f"data/{stock}.csv"
        df.to_csv(filename)

        print(f"\nSaved as: {filename}")

except Exception as e:
    print("An error occurred while downloading data.")
    print("Error:", e)
```

## Fit the Model Using scikit-learn

## LinearRegression

```python
import pandas as pd 
from sklearn.linear_model import LinearRegression

#remove unwanted ticker
df = df[df['Price'] != 'Ticker'].copy()

numeric_columns = ['Close', 'High', 'Low', 'Open', 'Volume']

for col in numeric_columns : 
    df[col] = pd.to_numeric(df[col], errors='coerce')

# remove missing value 
df = df.dropna()

# Sort data from oldest to newest
df['Price'] = pd.to_datetime(df['Price'])
df = df.sort_values('Price').reset_index(drop=True)

# Create tomorrow's target

df['Tomorrow_Open'] = df['Open'].shift(-1)
df['Tomorrow_High'] = df['High'].shift(-1)
df['Tomorrow_Low'] = df['Low'].shift(-1)
df['Tomorrow_Close'] = df['Close'].shift(-1)
df['Tomorrow_Volume'] = df['Volume'].shift(-1)

# feture
X = df[['Open', 'High', 'Low', 'Close', 'Volume']]

# Target
y = df[
    [
        'Tomorrow_Open',
        'Tomorrow_High',
        'Tomorrow_Low',
        'Tomorrow_Close',
        'Tomorrow_Volume'
    ]
]

# 5. Remove last row
# Last row has no tomorrow's actual value
X = X.iloc[:-1]
y = y.iloc[:-1]


# Create Model 
model = LinearRegression()

# Train Model
model.fit(X, y)
```

```python
latest_data = df[
                    ['Open', 'High', 'Low', 'Close', 'Volume']
                ].iloc[-1:]

prediction = model.predict(latest_data)

print("Tomorrow Open  :", prediction[0][0])
print("Tomorrow High  :", prediction[0][1])
print("Tomorrow Low   :", prediction[0][2])
print("Tomorrow Close :", prediction[0][3])
print("Tomorrow Volume:", prediction[0][4])
```
